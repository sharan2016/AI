```python
print("NAME:Sharan")
print("USN:1BM24CS367")
KB = [
    ["~Food(x)", "Likes(John,x)"],
    ["Food(Apple)"],
    ["Food(Vegetables)"],
    ["~Eats(x,y)", "Killed(x)", "Food(y)"],
    ["Eats(Anil,Peanuts)"],
    ["Alive(Anil)"],
    ["~Eats(Anil,x)", "Eats(Harry,x)"],
    ["~Alive(x)", "~Killed(x)"],
    ["Killed(x)", "Alive(x)"]
]

query = "Likes(John,Peanuts)"

def substitute(clause, substitutions):
    for var, value in substitutions.items():
        clause = clause.replace(var, value)
    return clause

def resolve(c1, c2):
    for lit1 in c1:
        for lit2 in c2:
            if lit1.startswith("~") and lit1[1:] == lit2:
                return [x for x in c1 if x != lit1] + [
                    x for x in c2 if x != lit2
                ]
            if lit2.startswith("~") and lit2[1:] == lit1:
                return [x for x in c1 if x != lit1] + [
                    x for x in c2 if x != lit2
                ]
    return None

print("KNOWLEDGE BASE")
for i, clause in enumerate(KB, 1):
    print("C" + str(i), ":", " OR ".join(clause))

print("\nQUERY:", query)

print("\nRESOLUTION STEPS")

c1 = ["Alive(Anil)"]
c2 = ["~Alive(Anil)", "~Killed(Anil)"]
r1 = resolve(c1, c2)
print("Step 1:", r1)

c3 = ["Eats(Anil,Peanuts)"]
c4 = ["~Eats(Anil,Peanuts)", "Killed(Anil)", "Food(Peanuts)"]
r2 = resolve(c3, c4)
print("Step 2:", r2)

r3 = resolve(r2, r1)
print("Step 3:", r3)

c5 = ["~Food(Peanuts)", "Likes(John,Peanuts)"]
r4 = resolve(r3, c5)
print("Step 4:", r4)

c6 = ["~Likes(John,Peanuts)"]
r5 = resolve(r4, c6)
print("Step 5:", r5)

if r5 == []:
    print("\nEMPTY CLAUSE OBTAINED")
    print("QUERY PROVED SUCCESSFULLY")
    print("John likes peanuts.")
else:
    print("\nQUERY NOT PROVED")
```
